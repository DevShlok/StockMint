```python
#TRAINING & TESTING DATA
data_training= pd.DataFrame(df['Close'][0:int(len(df)*0.70)])
data_testing= pd.DataFrame(df['Close'][int(len(df)*0.70):int(len(df))])
```

```python
data_training.shape
```

```python
data_testing.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
closing_price=df[['Close']]
scaler = MinMaxScaler (feature_range =(0,1))
data_training_array = scaler.fit_transform(data_training)
scaled_data= scaler.fit_transform(closing_price['Close'].dropna())
```

```python
data_training_array.shape
```

```python
data_training_array
```

```python
#LSTM MODEL
from keras.layers import Dense, LSTM, Dropout
from keras.models import Sequential
```

```python
# DATA FOR TESTING AND TRAINING
x_data=[]
y_data=[]
base_days=100

for i in range(base_days, len(scaled_data)):
    x_data.append(scaled_data[i-base_days:i])
    y_data.append(scaled_data[i])

x_data, y_data = np.array(x_data), np.array(y_data)
train_size= int(len(x_data)*0.9)
x_train, y_train = x_data[:train_size], y_data[:train_size]
x_test, y_test = x_data[train_size:], y_data[train_size:]
```

```python
model = Sequential ([
    LSTM(128, return_sequences=True, input_shape=(x_train.shape[1], 1)),
    LSTM(64, return_sequences= False), 
    Dense(25),
    Dense(1)           
])
model.compile(optimizer='adam', loss='mean_squared_error')
model.summary()
```

```python
model.summary()
```

```python

model.fit(x_train, y_train,batch_size=5 ,epochs=50)
```

```python
predictions = model.predict(x_test)
inv_predictions = scaler.inverse_transform(predictions)
inv_y_test = scaler.inverse_transform(y_test)
inv_predictions 
```

```python
plotting_data=pd.DataFrame({
    'Original': inv_y_test.flatten(),
    'Prediction': inv_predictions.flatten()
},
index=closing_price.index[train_size+base_days:]
)
plt.figure(figsize=(12,6))
plt.plot(plotting_data.index,plotting_data['Original'], label=f'{stock} Closing Price', linewidth=1, color='Red') 
plt.plot(plotting_data.index,plotting_data['Prediction'], linewidth=2, color='Green') 
plt.title(f'{stock} Predictions VS Original Price')
plt.xlabel('Years')
plt.ylabel('Price in USD')
plt.legend()
plt.show()
```

```python
past_100_days= scaled_data[-100:].reshape(1,-1,1)
future_predictions=[]
for _ in range (10):
    next_days=model.predict(past_100_days)
    future_predictions.append(scaler.inverse_transform(next_days))
    past_100_days=np.append(past_100_days[:,1:,:],next_days.reshape(1,1,-1),axis=1)
```

```python
future_predictions=np.array(future_predictions).flatten()
plt.figure(figsize=(12,6))
plt.plot(range(1,11),future_predictions,marker='o',label=f'{stock} Prediction of Future Prices', linewidth=1, color='Green') 
for i, value in enumerate(future_predictions):
    plt.text(i+1, value, f'{value:.2f}', fontsize=10, ha='center', va='bottom', color='black')

plt.title(f'{stock} Future Closing Price Predictions')
plt.xlabel('Days Ahead')
plt.ylabel('Price in USD')
plt.legend()
plt.show()
```
